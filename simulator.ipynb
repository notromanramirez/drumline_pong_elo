```python
import numpy as np

def create_tree(n):
    list_size = int(np.pow(2, np.ceil(np.log2(n))))
    ret_list = [ if (i//2) else (i + 1) // 2 for i in range(list_size)]
    return ret_list

# for i in range(1, 17):
#     print(i, create_tree(i))

my_list = list()
def compare_leaf(l):
    # l is a sub list
    print(l)
    if len(l) == 1:
        return l
    else:
        a = compare_leaf(l[:len(l)//2])
        b = compare_leaf(l[len(l)//2:])
        if sum(a) > sum(b):
            return a + b
        if sum(a) < sum(b):
            return b + a
        return l
    
compare_leaf(create_tree(8))

```
